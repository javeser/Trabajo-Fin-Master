```python
from rk_transport import FarRKTrkTransport, propagate_to_vertex

def propagate_to_vertex(vz, xi, yi, zi, pxi, pyi, pzi, qoverp,rk):
    xf,yf,txf,tyf = rk.propagate(vz/10., 1., xi/10., yi/10., zi/10., pxi/1000., pyi/1000., pzi/1000., qoverp*1000.)
    mom = np.array([txf,tyf,1])*((np.abs(1/(qoverp*1000.)))/np.sqrt(txf**2. + tyf**2. + 1.))
    return mom[0]*1000., mom[1]*1000., mom[2]*1000., xf*10., yf*10.

MagUp = True

rk = FarRKTrkTransport(MagUp)

vertex_values = pd.DataFrame(columns=['p_PX_V','p_PY_V','p_PZ_V','p_X_V','p_Y_V'])
vertex_values2 = pd.DataFrame(columns=['pim_PX_V','pim_PY_V','pim_PZ_V','pim_X_V','pim_Y_V'])

vertex_values[['p_PX_V','p_PY_V','p_PZ_V','p_X_V','p_Y_V']] = mc_data_selected.apply(lambda row: propagate_to_vertex(row.L_END_VZ,row.p_TRACK_POS_CLOSESTTOBEAM_X,row.p_TRACK_POS_CLOSESTTOBEAM_Y,row.p_TRACK_POS_CLOSESTTOBEAM_Z,row.p_PX,row.p_PY,row.p_PZ,row.p_QOVERP,rk),axis='columns',result_type='expand')
vertex_values2[['pim_PX_V','pim_PY_V','pim_PZ_V','pim_X_V','pim_Y_V']] = mc_data_selected.apply(lambda row: propagate_to_vertex(row.L_END_VZ,row.pim_TRACK_POS_CLOSESTTOBEAM_X,row.pim_TRACK_POS_CLOSESTTOBEAM_Y,row.pim_TRACK_POS_CLOSESTTOBEAM_Z,row.pim_PX,row.pim_PY,row.pim_PZ,row.pim_QOVERP,rk),axis='columns',result_type='expand')

mc_data_selected = pd.concat([mc_data_selected,vertex_values],axis='columns')
mc_data_selected = pd.concat([mc_data_selected,vertex_values2],axis='columns')
```

```python
import pandas as pd
from rk_doca import FarRKDOCA

# Supongamos que 'rk' está definido previamente
my_doca = FarRKDOCA(rk, 10)

def eval_units (x1,y1, z1, tx1, ty1, qoverp1,x2,y2, z2, tx2, ty2, qoverp2, my_doca):
    doca = my_doca.eval(x1/10.,y1/10., z1/10., tx1, ty1, qoverp1*1000., x2/10.,y2/10., z2/10., tx2, ty2, qoverp2*1000.)
    return doca*10.

# Calcula las nuevas columnas p_TX, p_TY, pim_TX, pim_TY
mc_data_selected['p_TX'] = mc_data_selected['p_PX'] / mc_data_selected['p_PZ']
mc_data_selected['p_TY'] = mc_data_selected['p_PY'] / mc_data_selected['p_PZ']
mc_data_selected['pim_TX'] = mc_data_selected['pim_PX'] / mc_data_selected['pim_PZ']
mc_data_selected['pim_TY'] = mc_data_selected['pim_PY'] / mc_data_selected['pim_PZ']

# DataFrame para almacenar los valores de DOCA
doca_value = pd.DataFrame(columns=['DOCA'])

# Calcula los valores de DOCA
doca_value['DOCA'] = mc_data_selected.apply(lambda row: eval_units(
    row.p_TRACK_POS_CLOSESTTOBEAM_X, row.p_TRACK_POS_CLOSESTTOBEAM_Y, row.p_TRACK_POS_CLOSESTTOBEAM_Z,
    row.p_TX, row.p_TY, row.p_QOVERP,
    row.pim_TRACK_POS_CLOSESTTOBEAM_X, row.pim_TRACK_POS_CLOSESTTOBEAM_Y, row.pim_TRACK_POS_CLOSESTTOBEAM_Z,
    row.pim_TX, row.pim_TY, row.pim_QOVERP, my_doca), axis=1)

# Concatenar los valores de DOCA al DataFrame original
mc_data_selected = pd.concat([mc_data_selected, doca_value], axis=1)


```

```python
# Create the figure and axes
fig, axes = plt.subplots()
binning = np.linspace(0, 100, 100)
density = False

# Plot histograms
hist1, bins1, _ = axes.hist(mc_data_selected.query('(L_BKGCAT ==0) ')['DOCA'], bins=binning, label='MC signal DOCA', color='green', alpha=0.7, histtype='step',density= density)


# Set labels and legend
plt.xlabel('DOCA [mm]')
plt.ylabel('Entries') 
plt.legend()

# Display the plot
plt.show()
```
